```latex
A trip covers $d = 2.4$ kilometers in $t = 12$ minutes.

$$	ext{speed in km/h} = rac{d}{t/60}$$
```

### How the filename reaches pandas
`pd.read_csv("filename.csv")` reads a CSV into a pandas DataFrame. The quoted filename is a path on the **runtime**. It is not a path to your laptop, and reading it does not change the original CSV.

| If you see... | Check this |
|---|---|
| `FileNotFoundError` | Is the runtime connected? Is the file uploaded? Does the name, including `.csv`, match exactly? |
| A name ending in `(1).csv` | Rename the runtime copy to the required filename. |
| The file inside a subfolder | Move/upload it to the top level shown by the Files panel; keep this lab's relative path. |
| A permission/login page instead of a table | You may have downloaded a webpage rather than the actual CSV. Download the course CSV again. |
| A file visible in Drive but absent from Files | Saving the notebook in Drive does not upload the CSV to the runtime. Upload the CSV separately. |
